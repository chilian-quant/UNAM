### Ejercicio 1: Calculadora Simple

```python
# Solicitar dos números al usuario
num1 = float(input("Ingresa el primer número: "))
num2 = float(input("Ingresa el segundo número: "))

# Realizar operaciones básicas
suma = num1 + num2
resta = num1 - num2
multiplicacion = num1 * num2
division = num1 / num2 if num2 != 0 else "Error (división indefinida)"

# Mostrar resultados
print(f"Resultados para {num1} y {num2}:")
print(f"Suma: {suma}")
print(f"Resta: {resta}")
print(f"Multiplicación: {multiplicacion}")
print(f"División: {division}")
```

### Ejercicio 2: Conversión de Temperaturas

```python
# Solicitar temperatura en Celsius
celsius = float(input("Ingresa la temperatura en grados Celsius: "))

# Fórmula de conversión: (C * 9/5) + 32
fahrenheit = (celsius * 9/5) + 32

print(f"{celsius}°C equivale a {fahrenheit}°F")
```

### Ejercicio 3: Cálculo del Área de un Triángulo

```python
# Solicitar base y altura del triángulo
base = float(input("Ingresa la base del triángulo: "))
altura = float(input("Ingresa la altura del triángulo: "))

# Calcular el área
area = (base * altura) / 2

print(f"El área del triángulo con base {base} y altura {altura} es: {area}")
```

EJERCICIO 4: Cambio de moneda

```python
# Solicitar la tasa de cambio actual (cuántos euros equivale a 1 dólar)
tasa_cambio = float(input("¿Cuál es el precio del euro en dólares actualmente? (ej. 1.09): "))

# Solicitar la cantidad en dólares a convertir
dolares = float(input("Ingresa la cantidad en dólares que deseas convertir: "))

# Calcular el equivalente en euros
euros = dolares / tasa_cambio

# Mostrar el resultado
print(f"{dolares} dólares equivalen a {euros:.2f} euros.")
```

### Ejercicio 5: Calculadora de Descuentos

```python
# Solicitar el precio original del producto
precio_original = float(input("Ingresa el precio original del producto: "))

# Solicitar el porcentaje de descuento
porcentaje_descuento = float(input("Ingresa el porcentaje de descuento (ej. 20 para 20%): "))

# Calcular el monto del descuento y el precio final
descuento = precio_original * (porcentaje_descuento / 100)
precio_final = precio_original - descuento

# Mostrar los resultados utilizando variables
print(f"\n--- Resumen de Compra ---")
print(f"Precio original: ${precio_original:.2f}")
print(f"Descuento aplicado ({porcentaje_descuento}%): -${descuento:.2f}")
print(f"Precio final a pagar: ${precio_final:.2f}")
```
