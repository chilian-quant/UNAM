```python
#Dado un triangul con lados de longitudes a=7, b=8 y c=9 escriba un programa de python que calcule e imprima el area del triangulo utilizando la formula de heron donde

#parte 1 numeros fijos
d= 7
e= 8
f= 9

semiperimetro_fijo = (d+e+f)/2
area = (semiperimetro_fijo*(semiperimetro_fijo-d)*(semiperimetro_fijo-e)*(semiperimetro_fijo-f))**0.5
print("El area del triangulo es: ", area)

#parte 2 cualesquiera numeros
a= float(input("Ingrese la longitud del lado a: "))
b= float(input("Ingrese la longitud del lado b: "))
c= float(input("Ingrese la longitud del lado c: "))

semiperimetro = (a+b+c)/2
area = (semiperimetro*(semiperimetro-a)*(semiperimetro-b)*(semiperimetro-c))**0.5
print("El area del triangulo es: ", area)




```

El Sistema Geodestico Mundial es un conjunto de reglas internacionales para describir la forma de la Tierra. En la utima revisio, WGS-84 el geoide terrestre se aproxima a un elipsoide de referencia que adopta la forma de un esferoide achatado con semiejes mayor y menos **a= 6378137.0** m **c=6356752.314245** m respectivamente. utilice la formula para el area de superficei de un esferoide oblato.

Para calcular el area de la superficie de este elipsode de referencia y comparelo con el area de la superfie de la tierra suponiendo que se trata de una esfera con radio r = **6371km**

```python
#El Sistema Geodestico Mundial es un conjunto de reglas internacionales para describir la forma de la Tierra. En la utima revisio, WGS-84 el geoide terrestre se aproxima a un elipsoide de referencia que adopta la forma de un esferoide achatado con semiejes mayor y menos **a= 6378137.0** m **c=6356752.314245** m respectivamente. utilice la formula para el area de superficei de un esferoide oblato. Para calcular el area de la superficie de este elipsode de referencia y comparelo con el area de la superfie de la tierra suponiendo que se trata de una esfera con radio r = **6371km**
import math
a= 6378137.0
c= 6356752.314245
e= (1-((c**2)/(a**2)))**0.5
r= 6371000

#Formula para el area de superficie de un esferoide oblato
s_obl= (2*(math.pi)*a**2)*(1+((1-e**2)/e)*(math.atanh(e)))

#Superficie de una esfera
se= 4*(math.pi)*(r**2)

#Diferencia de resultados
dif= ((se-s_obl)/se)*100

# Convertir a km^2
s_obl_km2 = s_obl / 1e6
se_km2 = se / 1e6

print(f"Área del Elipsoide (WGS-84): {s_obl_km2:,.2f} km²")
print(f"Área de la Esfera (r=6371km): {se_km2:,.2f} km²")
print(f"Diferencia porcentual:{math.fabs(dif):.6f}%")
```
