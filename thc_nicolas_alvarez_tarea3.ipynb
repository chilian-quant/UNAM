```python
#Ejercicio 1 Rectas. Programar un script que determine si 2 rectas dadas por las ecuaciones y1=m1x+b1 y y2=m2x+b2 se intersecan son paralelas o coincidentes.
import math

# Pedir al usuario los valores de las diferentes variables (usamos float para permitir decimales)
m1 = float(input("Ingrese el valor de m1: "))
b1 = float(input("Ingrese el valor de b1: "))
m2 = float(input("Ingrese el valor de m2: "))
b2 = float(input("Ingrese el valor de b2: "))

# Evaluación de las posiciones relativas de las dos rectas con indentación uniforme (4 espacios)
if m1 == m2:
    if b1 == b2:
        print(f"Las rectas son COINCIDENTES (son la misma recta porque m1 = m2 = {m1} y b1 = b2 = {b1}).")
    else:
        print(f"Las rectas son PARALELAS (tienen la misma pendiente {m1} pero se cruzan en diferentes puntos del eje Y: b1={b1} y b2={b2}).")
else:
    # Si las pendientes son diferentes, se intersecan en un punto único
    x_interseccion = (b2 - b1) / (m1 - m2)
    y_interseccion = m1 * x_interseccion + b1
    print(f"Las rectas se intersecan en un solo punto porque sus pendientes son distintas (m1={m1} y m2={m2}).")
    print(f"Punto de intersección: ({x_interseccion:.2f}, {y_interseccion:.2f})")
```

```python
#Ejercicio 2 Rectas Estimar el perimetro aprozimado (P) de una elipse usando la formula de ramujan.
#Declaras los semiejes a=8 y b=5, usando igualmente otro par de valores e imprimir resulatdo
import math

# Pares de prueba a evaluar: (a, b)
casos = [
    (8, 5),   # Par inicial propuesto
    (10, 3),  # Segundo par propuesto
    (6, 6)    # Caso a == b
]

for a, b in casos:
    if a > b:
        # Elipse
        p = math.pi * (3 * (a + b) - math.sqrt((3 * a + b) * (a + 3 * b)))
        print(f"Elipse (a={a}, b={b}):")
        print(f"  Perímetro aproximado = {p:.4f} unidades\n")

    elif a < b:
        # Elipse
        p = math.pi * (3 * (a + b) - math.sqrt((3 * a + b) * (a + 3 * b)))
        print(f"Elipse vertical (a={a}, b={b}):")
        print(f"  Perímetro aproximado = {p:.4f} unidades\n")

    else:
        # Caso a == b reducción exacta a una circunferencia
        r = a or b
        p = math.pi * (3 * (r + r) - math.sqrt((3 * r + r) * (r + 3 * r)))
        print(f"Caso especial a == b (a={a}, b={b}):")
        print("  La fórmula se reduce algebraicamente a P = 2*pi*r.")
        print(f"  Perímetro calculado (circunferencia de radio {r}) = {p:.4f} unidades")

```

```python
#Ejercicio desigualdad triangular y formula de Heron

while True:
    #Solicitar al usuario que ingrese los valores de los lados de un triangulo
    a = int(input("ingrese el valor del lado a: "))
    b = int(input("ingrese el valor del lado b: "))
    c = int(input("ingrese el valor del lado c: "))

    #Validar si en verdad es un triangulo (Desigualdad Triangular)
    if a + b > c and a + c > b and b + c > a :
        print(f"Como los lados {a}, {b}, {c} cumplen las 3 desigualdades si es un triangulo")

        # Si forman un triangulo indicar si es un Triangulo Equilatero, Isoceles o Escaleno
        if a == b == c :
            print(f"Como los 3 lados del Triangulo (a,b,c ={a}) son iguales entonces es un Triangulo Equilatero")
        elif a == b or a == c or b == c :
            print(f"Como alguno de los lados del Triangulo (a=b, a=c o b=c) son iguales entonces es un triangulo Isoceles")
        else :
            print(f"Como los 3 lados del Triangulo ({a},{b},{c}) son diferentes entonces es un Triangulo Escaleno")

        # Si forman un triangulo utilice la Formula de Heron de Alejandria para calcular el area del triangulo
        semiperimetro = (a + b + c) / 2
        area = (semiperimetro * (semiperimetro - a) * (semiperimetro - b) * (semiperimetro - c)) ** 0.5
        print(f"El semiperímetro es: {semiperimetro}")
        print(f"El área es: {area:.2f}")
    else:
        print(f"Alguna de las 3 desigualdades entre {a}, {b}, {c} no se cumple entonces no es un triangulo")

    # Preguntar al usuario si desea repetir el programa
    repetir = input("\n¿Desea analizar otro triángulo? (s/n): ")
    if repetir != 's':
        print("¡Programa finalizado!")
        break
```

```python
# Ejercicio 6. Energía mecánica total con toma de decisiones (if-elif-else)
import math

# Variables dadas
masa = 1200       # en kg
velocidad = 25    # en m/s
gravedad = 9.81   # en m/s^2
altura = 15       # en metros

# 1. Cálculo de las energías (Fórmula corregida utilizando el operador *)
Ec = (1/2) * masa * (velocidad**2)   # Energía Cinética
Ep = masa * gravedad * altura        # Energía Potencial
Em = Ec + Ep                         # Energía Mecánica Total

# Mostrar los resultados calculados
print("--- Resultados de Energía ---")
print(f"Energía Cinética (Ec): {Ec:.2f} J")
print(f"Energía Potencial (Ep): {Ep:.2f} J")
print(f"Energía Mecánica Total (Em): {Em:.2f} J\n")

print("--- Análisis del Estado del Objeto ---")
if velocidad == 0 and altura == 0:
    print("El objeto está en reposo absoluto a nivel del suelo (Sin energía).")

elif velocidad == 0:
    print(f"El objeto está en reposo a una altura de {altura}m. Toda su energía es Potencial ({Ep:.2f} J).")

elif altura == 0:
    print(f"El objeto se mueve a nivel del suelo. Toda su energía es Cinética ({Ec:.2f} J).")

else:
    print(f"El objeto posee movimiento y altura simultáneamente.")
```

```python
# Ejercicio 9. Volumen y Área de la Superficie de una Pirámide Regular Recta.
import math

#Solicitar al usuario las varibales para hacer los calculos
altura = int(input("Ingrese la altura de la piramide: "))
base = int(input("Ingrese el numero de lados de la piramide: "))
longitud = int(input("Ingrese la longitud de la base de la piramide: "))

if altura < 0 or base < 0 or longitud < 0 :
  print("ERROR, No se puede convertir un numero negativo")
else :
  # 1. Calcular variables necesarias para Volumen y Superficie
  # Calculo de la cotangente
  a = ((1/2)*longitud) * (1/math.tan(math.radians(base)))
  # Area de la base
  a_A = (1/2) * base * longitud * a

  #Apotema de la piramide (l)
  l = math.sqrt(altura**2 + a**2)

  #Calculo final para Volumen y Superficie
  Volumen = (1/3) * a_A * altura
  Superficie = a_A + ((1/2)*base*longitud*l)

  #RESULTADOS
  print("--- Resultados ---")
  print(f"El volumen de la piramide es: {Volumen:.2f}u^3")
  print(f"La superficie de la piramide es: {Superficie:.2f}u^2")

```

```python
# Ejercicio 10. Normalización de Ángulos y Determinación de Cuadrantes.
print("--- Normalización de Ángulos ---")
angulo = int(input("Ingrese el ángulo que desea normalizar: "))

# Guardamos el ángulo original para mostrarlo después
angulo_original = angulo

# Normalización del ángulo al intervalo [0, 360) utilizando if-elif-else
if angulo < -360:
    angulo = angulo + 720
elif angulo < 0:
    angulo = angulo + 360
elif angulo >= 720:
    angulo = angulo - 720
elif angulo >= 360:
    angulo = angulo - 360

# Determinación de ubicación (Ejes o Cuadrantes) con if-elif-else
if angulo == 0:
    ubicacion = "Sobre el Eje X positivo"
elif angulo == 90:
    ubicacion = "Sobre el Eje Y positivo"
elif angulo == 180:
    ubicacion = "Sobre el Eje X negativo"
elif angulo == 270:
    ubicacion = "Sobre el Eje Y negativo"
elif 0 < angulo < 90:
    ubicacion = "Cuadrante I"
elif 90 < angulo < 180:
    ubicacion = "Cuadrante II"
elif 180 < angulo < 270:
    ubicacion = "Cuadrante III"
elif 270 < angulo < 360:
    ubicacion = "Cuadrante IV"
else:
    ubicacion = "Ángulo fuera del rango esperado"

# Imprimir resultados en pantalla
print(f"\nÁngulo original: {angulo_original}°")
print(f"Ángulo reducido: {angulo}° ({ubicacion})")
```
