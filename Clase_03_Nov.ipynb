## Ciclo **While**

```python
#Ejemplo de Ciclo While
cuentas = 0
while cuentas < 6:
    print(cuentas)
    cuentas = cuentas + 1

#Ejemplo declaracion inicial
precio = 100

precio -= 20    # Aplica descuento: precio = 100 - 20 (80)
precio *= 1.16  # Agrega 16% IVA:  precio = 80 * 1.16 (92.8)
precio /= 2     # Divide entre 2:  precio = 92.8 / 2 (46.4)

print(f"Precio final: ${precio}")

#Ejemplo de White Not

x = 0
while not x == 5:
    x = x + 1
    print("x = %d" % x)


#Ejemplo de While-else
i = 1
while i < 6:
  print(i)
  i += 1
else:
  print("i is no longer less than 6")

# Generar una secuencia de Fibonacci
a, b = 0, 1
while b < 100:
    print(b)
    a, b = b, a + b

```

# Ciclo **For**


```python
#Ejemplo con una lista
fruits = ["apple", "banana", "cherry"]
for x in fruits:
  print(x)

#Ejemplo de Intervalo
for y in range(0,30,3): #Secuencia de numeros que va del 0 al 30 de 3 en 3
  print(y)

#Ejemplo:
for z in range (10):
  print(z)
```

# Sentencia **Break**

```python
fruits = ["apple", "banana", "cherry"]
for x in fruits:
  print(x)
  if x == "banana":
    break

i = 1
while i < 6:
  print(i)
  if i == 3:
    break
  i += 1
```

#Sentencia **Continue**

¿Cuándo y para qué se usa?

* Filtrar datos válidos: Omitir datos defectuosos o que no cumplen un criterio y seguir procesando el resto.

* Evitar código anidado: Reducir la cantidad de bloques if-else identados dentro de un bucle.

* Procesar selecciones: Omitir ceros, valores negativos o elementos nulos en un cálculo masivo.

```python
fruits = ["apple", "banana", "cherry"]
for x in fruits:
  if x == "cherry":
    continue
  print(x)

i = 0
while i < 6:
  i += 1
  if i == 2:
    continue
  print(i)

```

# **EJERCICIOS CICLO WHILE**
Ejemplos 1, 3, 6


```python
#Ejericio 1 Calcular la suma de los números impares del 1 al 50.
si = 0
num = 1

while num <= 50:
    if num % 2 != 0:
        si += num
    num += 1
print("La suma de los números impares del 1 al 50 es:", si)
```

```python
#Ejercicio 3 Imprimir los múltiplos de 3 menores que 100 utilizando un bucle while.
num = 1
while num < 100:
    if num % 3 == 0:
        print("mult. de 3:", num)
    num += 1
```

```python
#Ejercicio 6 Imprimir los números primos menores que 100 utilizando un bucle while.
num = 2
while num < 100:
    primo = True
    div = 2
    while div * div <= num:
        if num % div == 0:
            primo = False
            break
        div += 1
    if primo:
        print(num)
    num += 1
```

```python
#Ejericio 5 Calcular el factorial de un número dado.
num = int(input("Ingrese un número: "))
factorial = 1
for i in range(1, num + 1):
    factorial *= i
print("El factorial de", num, "es:", factorial)
```
